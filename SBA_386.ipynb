```python
# Grab the tools
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the WooCommerce DataFrame
df = pd.read_csv('data/woocommerce-product-export.csv')
```

```python
# Display what columns exist
df.info
```

```python
# Generate summary (mean, min, max, etc.)
df.describe
```

```python
# Check for any missing/null values across all columns
df.isnull().sum()
```

```python
# Display the first 5 rows
df.head()
```

```python
# Display the last 5 rows
df.tail()
```

```python
# Filter, only show month numbers and total profit columns
df[['month_number', 'total_profit']]
```

```python
# Create a bar chart titled company profit per month
plt.bar(df["month_number"], df["total_profit"])

# Add labels, title, and grid lines
plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company Profit per Month")

plt.grid(True, linestyle="--")

plt.show()
```

```python
# Plot profit data w/ custom styling
plt.plot(
    df["month_number"],
    df["total_profit"],
    label="Profit data of last year",
    color="red",  # red dotted line
    linestyle=":",
    linewidth=3,
    marker="o",  # circular markers
    markerfacecolor="black",
)

# Set axis labels and title
plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

# Add a legend in the lower right corner
plt.legend(loc="lower right")

plt.show()
```

```python
# List of product categories we need to loop through
products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

# Loop through each product to plot its individual monthly sales data
for product in products:
    plt.plot(
        df['month_number'],
        df[product],
        label=f'{product} Sales Data',
        marker='o',
        linewidth=2,
    )

# Add axis labels, title, and legend
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales Data')
plt.legend(loc='upper left')

plt.show()
```

```python
# Create a scatter plot for bathing soap sales data per month
plt.scatter(df['month_number'], df['bathingsoap'], label='bathing soap Sales data')

# Configure grid, axis labels, title, and legend
plt.grid(True, linestyle="--")

plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.title('bathingsoap Sales data')
plt.legend(loc='upper left')

plt.show()
```

```python
# Define sample date and temperature values
date = ["25/12", "26/12", "27/12"]
temp = [8.5, 10.5, 6.8]
```

```python
# Built a custom-styled figure and axis with a dark background theme
fig1, ax1 = plt.subplots(figsize=(8, 5), facecolor='#121212')
ax1.set_facecolor('#1a1a1a')

# Plot the temperature trends with square markers and custom marker styling
ax1.plot(
    date,
    temp,
    color='#E5E7EB',
    linewidth=2.5,
    marker='s',
    markersize=8,
    markerfacecolor='#121212',
    markeredgecolor='#E5E7EB',
)

# Customize labels, title, spines (borders), ticks, and grid lines
ax1.set_xlabel('Date', fontsize=12, color='#D1D5DB', labelpad=10)
ax1.set_ylabel('temperature', fontsize=12, color='#D1D5DB', labelpad=10)
ax1.set_title(
    'Date-Wise Temperature', fontsize=15, fontweight='bold', color='#FFFFFF', pad=15
)

ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['left'].set_color('#4B5563')
ax1.spines['bottom'].set_color('#4B5563')
ax1.tick_params(colors='#9CA3AF')
ax1.grid(color='#374151', linestyle='--', alpha=0.5)

plt.show()
```

```python
# Define sample height and weight data points
height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]
```

```python
# Create a 2nd figure and axis for height versus weight visualization
fig2, ax2 = plt.subplots(figsize=(8, 5), facecolor='#121212')
ax2.set_facecolor('#1a1a1a')

# Plot average height with respect to weight using custom styling
ax2.plot(
    weight,
    height,
    label='Avg Height vs Weight',
    color='#39FF14',
    linestyle='-.',
    linewidth=2,
    marker='o',
    markersize=10,
    markerfacecolor='#39FF14',
    markeredgecolor='#121212',
    markeredgewidth=1.5,
)

# Customize axis labels, title, and theme colors
ax2.set_xlabel('Weight in kg.', fontsize=12, color='#D1D5DB', labelpad=10)
ax2.set_ylabel('Height in cm.', fontsize=12, color='#D1D5DB', labelpad=10)
ax2.set_title(
    'Average Weight with respect to Average Height.',
    fontsize=14,
    fontweight='bold',
    color='#FFFFFF',
    pad=15,
)

# Configure the plot legend with a custom background and border color
legend = ax2.legend(
    loc='lower right', facecolor='#1a1a1a', edgecolor='#4B5563', labelcolor='#D1D5DB'
)

# Adjust spines, tick colors, and grid lines
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_color('#4B5563')
ax2.spines['bottom'].set_color('#4B5563')
ax2.tick_params(colors='#9CA3AF')
ax2.grid(color='#374151', linestyle=':', alpha=0.6)

plt.show()
```
